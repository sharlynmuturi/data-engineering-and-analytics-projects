```python
import pandas as pd
import json
from sqlalchemy import create_engine
```

### Reading the large JSON Lines (JSONL) file line by line, parsing each JSON object and converting the data into a Pandas DataFrame for analysis.

```python
path = r'dataset/yelp_academic_dataset_business.json'

business_data = []
with open(path, 'r', encoding='utf-8') as f:
    for line in f:
        business_data.append(json.loads(line))

business_df = pd.DataFrame(business_data)
```

```python
path = r'dataset/yelp_academic_dataset_checkin.json'

checkin_data = []
with open(path, 'r', encoding='utf-8') as f:
    for line in f:
        checkin_data.append(json.loads(line))

checkin_df = pd.DataFrame(checkin_data)
```

```python
path = r'dataset/yelp_academic_dataset_review.json'

review_data = []
with open(path, 'r', encoding='utf-8') as f:
    for line in f:
        review_data.append(json.loads(line))

review_df = pd.DataFrame(review_data)
```

```python
path = r'dataset/yelp_academic_dataset_tip.json'

tip_data = []
with open(path, 'r', encoding='utf-8') as f:
    for line in f:
        tip_data.append(json.loads(line))

tip_df = pd.DataFrame(tip_data)
```

```python
path = r'dataset/yelp_academic_dataset_user.json'

user_data = []
with open(path, 'r', encoding='utf-8') as f:
    for line in f:
        user_data.append(json.loads(line))

user_df = pd.DataFrame(user_data)
```

```python
print(business_df.shape)
print(checkin_df.shape)
print(review_df.shape)
print(tip_df.shape)
print(user_df.shape)
```

### Removing nested JSON columns to simplify analysis

```python
business_df.drop(['attributes', 'hours'], axis = 1, inplace = True)
```

### Loading each DataFrame into a SQL table in chunks

```python
# taking 5% or 50k rows for the large dataset size
business_sample = business_df.sample(frac=0.1, random_state=42)
checkin_sample  = checkin_df.sample(frac=0.1, random_state=42)
review_sample   = review_df.sample(50000, random_state=42)
tip_sample      = tip_df.sample(50000, random_state=42)
user_sample     = user_df.sample(50000, random_state=42)
```

```python
engine = create_engine('sqlite:///yelp.db')

def load_dataframe(df, table_name, engine, chunksize=10000):
    df.to_sql(table_name, engine, if_exists='replace', index=False, chunksize=chunksize)

# Loading sampled data
load_dataframe(business_sample, 'business', engine)
load_dataframe(checkin_sample, 'checkin', engine)
load_dataframe(review_sample, 'review', engine)
load_dataframe(tip_sample, 'tip', engine)
load_dataframe(user_sample, 'user', engine)
```
